```python
# importing essential libs 

import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from keras.models import Sequential
from keras.layers import *
```

```python
# sourcing synthetic data from kaagle or similar website

df=pd.read_csv('energy_consumption.csv',index_col=['Datetime'],parse_dates=['Datetime'])
df.head()
```

```python
df.info()
```

```python
# split the data into the train and split

train_size=int(len(df)*0.8)
length_df=len(df)

train=df[0:train_size]
test=df[train_size:length_df]
```

```python
print("len_train: ",len(train), "len_test: ", len(test))
```

```python
# scale the values to fit into min - max range 
# because it will fit into the specific range 

scaler=MinMaxScaler()
scaled_train=scaler.fit_transform(train)
```

```python
#scaling the test and train data

time_step=50

x_train=[]
y_train=[]

for i in range(time_step,len(scaled_train)):
    x_train.append(scaled_train[i-time_step:i,0])
    y_train.append(scaled_train[i,0])
```

```python
#convert the list into numpy array
x_train=np.array(x_train)
y_train=np.array(y_train)
```

```python
x_train.shape
```

```python
#reshaping to add another dim 
x_train=np.reshape(x_train,(x_train.shape[0],x_train.shape[1],1))
x_train.shape
```

```python
#construct the sequential lstm model to capture information and make prediction out of it 

lstm_model=Sequential()

lstm_model.add(LSTM(64,activation='tanh',return_sequences=True,input_shape=(x_train.shape[1],1)))
lstm_model.add(Dropout(0.2)) # dropout used to model can understand the missing neuron to overcome overfitting

lstm_model.add(LSTM(64,activation='tanh',return_sequences=True))
lstm_model.add(Dropout(0.20))

lstm_model.add(LSTM(64,activation='tanh',return_sequences=True))
lstm_model.add(Dropout(0.20))

lstm_model.add(LSTM(64,activation='tanh',return_sequences=True))
lstm_model.add(Dropout(0.20))

lstm_model.add(LSTM(64,activation='tanh',return_sequences=True))
lstm_model.add(Dropout(0.20))

lstm_model.add(LSTM(64))
lstm_model.add(Dropout(0.20))

lstm_model.add(Dense(1))
```

```python
#concat all data
test_array=test.values
total_data=pd.concat((df['DOM_MW'],test['DOM_MW']),axis=0)
```

```python
inputs=total_data[len(total_data)-len(test)-time_step:].values.reshape(-1,1)
inputs=scaler.transform(inputs)
inputs.shape
```

```python
# creating test list
x_test=[]
for i in range(time_step,inputs.shape[0]):
    x_test.append(inputs[i-time_step:i-0])
x_test=np.array(x_test)
x_test.shape
```

```python
#model compilization using loss function and optimizer 
lstm_model.compile(optimizer='adam',loss='mse')
lstm_model.fit(x_train,y_train,epochs=5)
```

```python
# model prediction
y_pred=lstm_model.predict(x_test)
y_pred=scaler.inverse_transform(y_pred)
```

```python
#plotting the predicted and actual values 

plt.figure(figsize=(15,8))
plt.plot(test_array,color='red',label='Actual Energy')
plt.plot(y_pred,color='blue',label='Predicate Energy')
plt.xlabel('Time')
plt.ylabel('Energy')
plt.title("Actual Energy & Predicate Energy")
plt.legend()
plt.show()
```
